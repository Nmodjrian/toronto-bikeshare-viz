# Explore June 2024 ridership

Uses the officially downloaded June 2024 dataset. No cycling paths are inferred here.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
PROJECT = Path.cwd()
if PROJECT.name == 'notebooks':
    PROJECT = PROJECT.parent
DATA = PROJECT / 'data/processed/bikeshare_2024_06.parquet'
assert DATA.exists(), f'Missing {DATA}; run the download and prepare scripts from the repository root.'
df = pd.read_parquet(DATA)
df['Start_Time'] = pd.to_datetime(df['Start_Time'], errors='coerce')
print(f'Loaded {len(df):,} records')
```

```python
daily=df.groupby(df['Start_Time'].dt.date).size()
fig,ax=plt.subplots(figsize=(13,5))
ax.plot(pd.to_datetime(daily.index),daily.values,marker='o')
ax.set(title='Bike Share Toronto | Daily trips | June 2024',ylabel='Recorded trips')
ax.grid(alpha=.2); fig.autofmt_xdate();plt.show()
```

```python
fig,axes=plt.subplots(1,2,figsize=(12,4))
df['User_Type'].fillna('Unknown').value_counts().plot.bar(ax=axes[0],title='User types')
(pd.to_numeric(df['Trip_Duration'],errors='coerce')/60).clip(0,90).hist(bins=45,ax=axes[1])
axes[1].set(title='Recorded duration (clipped at 90 min)',xlabel='Minutes');plt.tight_layout();plt.show()
```
